### Part A

```python
# A1
fc = np.array([23.4, 25.1, 19.8, 22.6, 24.0, 20.5, 26.3, 22.9])
print(fc)
print(type(fc))
```

```python
# A2
print('mean:', np.mean(fc))
print('std :', np.std(fc))
print('max :', np.max(fc))
print('min :', np.min(fc))
```

```python
# A3 — the scalar 10.197 is "broadcast" to every element
fc_kg = fc * 10.197
print(np.round(fc_kg, 1))
```

```python
# A4 — negative = the cylinder is BELOW the design strength
diff = fc - 21
print(diff)
```

```python
# A5
fails = fc < 21
print(fails)
print('Number of failed cylinders:', np.sum(fails))
```

```python
# A6
print([23.4, 25.1] + [19.8, 22.6])                      # lists: concatenation (joins them)
print(np.array([23.4, 25.1]) + np.array([19.8, 22.6]))  # arrays: element-wise addition
```

**A6 answer:** With lists, `+` **joins** the two lists. With NumPy arrays, `+` **adds** element by element.

### Part B

```python
settle = np.array([0.0, 1.2, 2.1, 2.9, 3.5, 4.0, 4.4, 4.7, 5.0, 5.2, 5.3, 5.5, 5.6, 5.6])
print('B1:', settle[3])
print('B2:', settle[-1])
print('B3:', settle[0:7])     # or settle[:7] — index 7 is NOT included
print('B4:', settle[::2])
print('B5:', settle[-3:])
print('B6:', settle[[1, 7, 13]])
```

```python
# B7 — settle[1:] is "tomorrow", settle[:-1] is "today"
daily = settle[1:] - settle[:-1]
print(np.round(daily, 1))
# The settlement gets slower over time: the soil is consolidating.
```

### Part E

```python
rain = np.array([110, 125, 118, 60, 20, 8, 5, 10, 25, 55, 70, 95])

# E1
months = np.arange(1, 13)
print('E1:', months)

# E2
q = rain.reshape(4, -1)
print('E2:\n', q)
print('Total per quarter:', np.sum(q, axis=1))

# E3
print('E3:\n', q.T, '\nshape:', q.T.shape)

# E4
print('E4 dry months:', months[rain < 30])

# E5
print(np.zeros((3, 4)))
print(np.ones((2, 2)))
```

### Part F

```python
# F1 — flip, running sum down the rows, flip back
acc = np.cumsum(loads[::-1], axis=0)[::-1]
print(acc)
```

```python
# F2 — area has shape (4,), acc has shape (3, 4).
# NumPy treats area as (1, 4) and repeats it for the 3 rows (broadcasting).
area = np.array([0.09, 0.12, 0.12, 0.09])
stress = acc / area
print(np.round(stress))
```

```python
# F3
over = stress > 8600
print(over)
print('Positions over the limit:', np.sum(over))
```

```python
# F4
print('Max stress per column:', np.round(np.max(stress, axis=0)))
print('Floor index (0 = Floor 1):', np.argmax(stress, axis=0))
```

**Engineering note:** the bottom floor always has the maximum stress, because it carries the whole building. In this example, on Floor 1, columns C1, C2 and C4 go over the limit. C1 is the worst because it has a small area. A designer would make these columns bigger or use stronger concrete. 🏗️
